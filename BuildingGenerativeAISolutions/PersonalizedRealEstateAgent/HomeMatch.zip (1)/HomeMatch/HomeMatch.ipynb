### Load the information from csv file, create embeddings and store it in vector database

```python
# let us load all langchain, openai and other stuff
from langchain.document_loaders.csv_loader import CSVLoader
from langchain.embeddings.openai import OpenAIEmbeddings
from langchain.llms import OpenAI
from langchain.text_splitter import CharacterTextSplitter
from langchain.vectorstores import Chroma
from langchain.chains import RetrievalQA
from langchain.prompts import PromptTemplate
from langchain import LLMChain
from langchain.chains.question_answering import load_qa_chain
```

```python
loader = CSVLoader(file_path='./listings_1.csv')
docs = loader.load()
docs
```

```python
model_name = 'gpt-3.5-turbo'
llm = OpenAI(model_name=model_name, temperature=0, max_tokens=1000, openai_api_key="YOUR_API_KEY")
```

```python
embeddings = OpenAIEmbeddings(openai_api_key="YOUR_API_KEY")
```

```python
db = Chroma.from_documents(docs, embeddings)
print(db._collection.count())
```

```python
db
```

```python
q1 =  "What is your total budget for house purchase?"
q2 = "How many bedrooms you need in the house?"
q3 = "How many bathrooms you need in the house?"
q4 = "What is the house size you are looking for?"
q5 = "Any other environmental, transportational or any other preference you have. To skip please enter NA"
q_list = [q1, q2, q3, q4 , q5]
q_list
```

```python
def ask_questions_query(q_list):
    """
    This function takes input from user and creates and returns query utilizing user responses
    """
    budget = input(q_list[0])
    br = input(q_list[1])
    bt = input(q_list[2])
    hs = input(q_list[3])
    misc = input(q_list[4])
    query = f'A house with approx total size of  {hs} sqft. Number of bedrooms and bathrooms in the house is  {br} and  {bt} respectively. Price of the house is approx {budget}. Other requirements are {misc}'
    return query
```

```python
def semantic_search(query):
    """
    This function takes query and return top 3 similar searches from vector database
    """
    similar_docs = db.similarity_search(query, k=2)
    return similar_docs
```

```python
def gen_aug_listing(similar_docs, template, query):
    """
    This function takes matching document, template and query
    and created augmented listings
    """
    for ind, context in enumerate(similar_docs):
        temperature = 0.0
        model = "gpt-3.5-turbo-instruct"
        response = openai.Completion.create(
            model = model,
            prompt = template.format(query=query, context=context),
            max_tokens = 1000,
            temperature=temperature            
            )
        output = response['choices'][0]['text'].strip()
        print(f'\n\n ######### Listing {ind} ##########')
        print(output)
```

```python
def run_app(template):
    """
    This function creates query based on user inputs
    Runs similarity Search
    Generates augmented listings
    using respective functions
    """
    query = ask_questions_query(q_list)
    print(f' ##\n Query: {query} \n ###')
    similar_docs = semantic_search(query)
    print(f'printing similar docs ##### {similar_docs} ######')
    print("########### AUGMENTED LISTINGS BASED ON USER INPUT ########")
    gen_aug_listing(similar_docs, template, query)
```

```python
template = """
Personalize the following real estate listing description as given in Context.
Please do not alter the information on number of bedrooms, bathrooms, price and house size given in Context
Context: {context}


Buyer Preferences: {query}
"""
```

```python
print(template)
```

```python
run_app(template)
```
